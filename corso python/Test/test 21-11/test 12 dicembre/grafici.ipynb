```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re

# Librerie per il Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder
```

```python
#caricamento del dataset
df = pd.read_csv('cleaned_manga_dataset.csv')
df.head()
```

# Creazione dei grafici

```python
# Impostiamo lo stile
sns.set_theme(style="whitegrid")
plt.figure(figsize=(18, 10))

#Top 10 Manga per Vendite
plt.subplot(2, 2, 1)
top_10 = df.sort_values('Approximate sales in million(s)', ascending=False).head(10)
sns.barplot(data=top_10, x='Approximate sales in million(s)', y='Manga series', palette='viridis')
plt.title('Top 10 Manga Best-Seller')
plt.xlabel('Vendite (Milioni)')
```

# Distribuzione Demografica (Pie Chart)


```python

plt.figure(figsize=(30, 20))
plt.subplot(2, 2, 2)
demo_counts = df['Demographic'].value_counts()
plt.pie(demo_counts, labels=demo_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribuzione Demografica')
```

# Relazione Volumi vs Vendite (Scatter Plot)


```python

plt.figure(figsize=(35, 30))
plt.subplot(2, 2, 3)
sns.scatterplot(data=df, x='No. of collected volumes', y='Approximate sales in million(s)', 
                hue='Demographic', alpha=0.7, size='Approximate sales in million(s)')
plt.title('Relazione: Numero Volumi vs Vendite')
```

# Andamento temporale (Line Plot)


```python

plt.figure(figsize=(30, 25))
plt.subplot(2, 2, 4)
sales_by_year = df.groupby('Start_Year')['Approximate sales in million(s)'].mean().reset_index()
sns.lineplot(data=sales_by_year, x='Start_Year', y='Approximate sales in million(s)', color='coral')
plt.title('Media Vendite per Anno di Inizio Pubblicazione')
plt.ylabel('Media Vendite (Milioni)')
```
