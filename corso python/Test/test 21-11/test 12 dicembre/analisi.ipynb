```python
# colonna delle vendite sia numerica
df['Approximate sales in million(s)'] = pd.to_numeric(df['Approximate sales in million(s)'], errors='coerce')
print(df['Approximate sales in million(s)'].dtype)
df['No. of collected volumes'] = pd.to_numeric(df['No. of collected volumes'], errors='coerce')
print(df['No. of collected volumes'].dtype)

```

```python

# Ordiniamo il dataset per vendite (dal più alto al più basso) e prendiamo i primi 15
top_manga = df.sort_values('Approximate sales in million(s)', ascending=False).head(15)
print(top_manga)
```

```python
#pulizia dei dati
#facciamo in modo di rimuovere le righe con valori mancanti nelle colonne
df = df.dropna(subset=['Approximate sales in million(s)', 'No. of collected volumes'])
df = df.reset_index(drop=True)
print(df.info())
print(df.head())
```

```python
# Funzione per pulire e convertire colonne numeriche (rimuove errori)
def clean_currency(x):
    if isinstance(x, str):
        return pd.to_numeric(x.replace(',', ''), errors='coerce')
    return x
```

```python
# Convertiamo le colonne chiave in numeri
df['Approximate sales in million(s)'] = df['Approximate sales in million(s)'].apply(clean_currency)
df['No. of collected volumes'] = df['No. of collected volumes'].apply(clean_currency)
print(df.info())
```

```python
df.head()
```

```python
# Feature Engineering: Estraiamo l'anno di inizio dalla colonna "Serialized" (es. "1997-present" -> 1997)
def extract_year(text):
    match = re.search(r'(\d{4})', str(text))
    return int(match.group(1)) if match else np.nan

df['Start_Year'] = df['Serialized'].apply(extract_year)
```

```python

# Gestione valori mancanti (riempiamo l'anno con la mediana e rimuoviamo righe senza vendite)
df['Start_Year'] = df['Start_Year'].fillna(df['Start_Year'].median())
df.dropna(subset=['Approximate sales in million(s)', 'No. of collected volumes'], inplace=True)
```

```python
print(f"Dataset pulito: {df.shape[0]} righe rimanenti.")
print(df.head())
```

```python
#caricamento dataset pulito
df.to_csv('cleaned_manga_dataset.csv', index=False)
```
